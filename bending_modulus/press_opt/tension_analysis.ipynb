```python
import numpy as np
import os
import matplotlib.pyplot as plt
import tqdm
```

```python
def running_avg(lst):

    avg_lst = []
    std_lst = []

    for i in tqdm.tqdm(range(len(lst))):

       avg_lst.append(np.mean(lst[:i+1])) 
       std_lst.append(np.std(lst[:i+1]))

    avg_lst, std_lst = np.array(avg_lst), np.array(std_lst)

    plt.plot(avg_lst)
    plt.plot(lst, alpha = 0.3)
    plt.show()
```

```python
dir = "/home/gabriel/Dokumente/bachelor_thesis/dpd_simulation/bending_modulus/"

pressures = ["77", "76.2", "75", "72", "60", "80"]

for p in pressures:

    tension_data = np.loadtxt(os.path.join(dir, f"press_opt/tension_profile_{p}.dat"), skiprows = 2, usecols = 1)

    print(f"------------- p = {p} ---------------")

    print(f"lateral tension {np.mean(tension_data)} +/- {np.std(tension_data, ddof=1)}")

    running_avg(tension_data)
```
